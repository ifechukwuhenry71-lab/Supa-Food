```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv(r"C:\Users\User\Desktop\supa-foods-app\dataset.csv.csv")

df.head()
```

```python
df_copy= df.copy()
```

```python
print(df_copy.isnull().sum())
```

```python
df_copy["Main_Ingredients"] = df_copy["Main_Ingredients"].astype(str).str.lower()

df_copy.head()
```

```python
df_copy["clean_text"]= df_copy["Main_Ingredients"].str.replace(","," ")
df_copy["Tokenized_Ingredients"]= df_copy["clean_text"].str.split()

df_copy.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df_copy["clean_text"])


def recommend_by_similarity(user_input_ingredients, top_n=3):
    # Vectorize user input
    user_vec = tfidf.transform([user_input_ingredients.lower()])

    # Compute similarity between user input and all dataset dishes
    similarity_scores = cosine_similarity(user_vec, tfidf_matrix).flatten()

    # Get top matching indices
    top_indices = similarity_scores.argsort()[-top_n:][::-1]

    return df.iloc[top_indices][["Food_Name", "Description", "Main_Ingredients"]]
```
